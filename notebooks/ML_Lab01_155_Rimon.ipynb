```python
import numpy as np
np.random.seed(42)
marks = np.array([55, 72, 48, 90, 66, 81])
print('Marks:', marks)
print('Mean:', marks.mean())
print('Highest:', marks.max())
print('Passed:', marks[marks >= 50])
#tasks
# Minimum mark and standard deviation
print('Minimum:', marks.min())
print('Standard Deviation:', marks.std())

# Adding 5 bonus marks without changing the original array
bonus_marks = marks + 5
print('Bonus Marks:', bonus_marks)
print('Original Marks:', marks)

# Counting how many students passed
passed_count = np.sum(marks >= 50)
print('Number of students passed:', passed_count)

# Creating a 2 × 3 matrix
matrix = marks.reshape(2, 3)
print('2 × 3 Matrix:')
print(matrix)
```

```python
import pandas as pd
data = {
'student': ['Asha', 'Babu', 'Chayan', 'Dipa', 'Emon', 'Farah'],
'study_hours': [2, 5, 1, 7, 3, 6],
'sleep_hours': [8, 6, 9, 5, 7, 6],
'attendance': [70, 88, 55, 95, 75, 91],
'snacks': [3, 2, 5, 1, 4, 2],
'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}
df = pd.DataFrame(data)
display(df)
print(df.info())
print(df.describe())

#tasks
#Features and label
#The features are the columns used to predict the outcome:
#study_hours,sleep_hours,attendance,snacks
#The label/target is:passed

#Number of samples and columns
print("Samples:", df.shape[0])
print("Columns:", df.shape[1])

#Students with attendance of at least 80
high_attendance = df[df['attendance'] >= 80]
display(high_attendance)

#Average study hours for students who passed
average_study = df[df['passed'] == 'Yes']['study_hours'].mean()
print("Average study hours:", average_study)

#Sorting by attendance in descending order
sorted_df = df.sort_values(by='attendance', ascending=False)
display(sorted_df)

#Do snacks cause failure?
#No. This tiny dataset does not prove that snack purchases cause failure.
#The dataset shows a possible relationship between snack purchases and passing/failing, but it is far too small and uncontrolled to conclude that snacks cause failure.
```

```python
import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)
    return np.random.randint(40, 101, size=5)

print('Seeded run 1:', generate_scores(True))
print('Seeded run 2:', generate_scores(True))
print('Unseeded run 1:', generate_scores(False))
print('Unseeded run 2:', generate_scores(False))
```

**Tasks**

**Observation:**
Seeded run 1 and Seeded run 2 give identical outputs every time the cell is run, because setting a seed fixes the random number generator's starting point. Unseeded run 1 and Unseeded run 2 differ from each other, and also differ each time the cell is rerun, since no seed is set.

**If two groups claim different
accuracies because they used different random splits, can we fairly compare their models?**

No, not fairly. If two groups used different (unseeded) random splits, a difference in accuracy could simply be due to one group getting a luckier or easier test split, rather than their model actually being better. To fairly compare, both groups should use the same seed for their split, or use cross-validation and report average performance across multiple splits. Otherwise the comparison is affected by random chance, not just model quality.
